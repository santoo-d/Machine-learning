## Naïve Bayes Algorithm Implementation

This section will implement the Naïve Bayes algorithm from scratch. Since no dataset was provided, a simple synthetic dataset will be created first.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix, accuracy_score
import numpy as np

# 1. Create a simple synthetic dataset
np.random.seed(42)
X = np.random.rand(100, 2) * 10  # 100 samples, 2 features
y = (X[:, 0] + X[:, 1] > 10).astype(int) # Binary classification target

df = pd.DataFrame(X, columns=['feature_1', 'feature_2'])
df['target'] = y

print("Sample Dataset Head:")
display(df.head())
print("\nTarget Distribution:")
display(df['target'].value_counts())
```

### Data Splitting

Now, let's split the dataset into training and testing sets to evaluate the model's performance on unseen data.

```python
# 2. Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print(f"Training set size: {len(X_train)} samples")
print(f"Testing set size: {len(X_test)} samples")
```

### Naïve Bayes Model Training

We will use `GaussianNB` from `sklearn.naive_bayes`, which is suitable for normally distributed continuous data.

```python
# 3. Implement Naïve Bayes algorithm
classifier = GaussianNB()
classifier.fit(X_train, y_train)

print("Naïve Bayes Classifier trained successfully.")
```

### Predictions and Evaluation

Finally, we'll make predictions on the test set and evaluate the model using a confusion matrix and accuracy score.

```python
# 4. Make predictions on the test set
y_pred = classifier.predict(X_test)

# 5. Display the results using confusion matrix and accuracy
conf_matrix = confusion_matrix(y_test, y_pred)
accuracy = accuracy_score(y_test, y_pred)

print("Confusion Matrix:")
display(pd.DataFrame(conf_matrix,
                     index=['Actual Negative', 'Actual Positive'],
                     columns=['Predicted Negative', 'Predicted Positive']))

print(f"\nAccuracy: {accuracy:.2f}")
```
