```python
# Step 1: Import required libraries
import numpy as np
import pandas as pd
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Step 2: Load Iris dataset
iris = datasets.load_iris()

# Features (X) and Target (y)
X = iris.data
y = iris.target

# Optional: Convert to DataFrame for better understanding
df = pd.DataFrame(X, columns=iris.feature_names)
df['target'] = y
print("Dataset Preview:\n", df.head())

# Step 3: Split dataset into training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Step 4: Create Naive Bayes model
model = GaussianNB()

# Step 5: Train the model
model.fit(X_train, y_train)

# Step 6: Make predictions
y_pred = model.predict(X_test)

# Step 7: Evaluate the model
print("\nAccuracy:", accuracy_score(y_test, y_pred))

print("\nClassification Report:\n", classification_report(y_test, y_pred))

print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))

# Step 8: Test with a sample input
sample = [[5.1, 3.5, 1.4, 0.2]]  # Example: Setosa
prediction = model.predict(sample)
print("\nSample Prediction:", iris.target_names[prediction][0])
```
