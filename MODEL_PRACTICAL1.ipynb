```python
import numpy as np
from sklearn.mixture import GaussianMixture
import matplotlib.pyplot as plt

# Data
X = np.array([[1], [1.5], [2], [2.5], [3],[3.5],[4],[4.5],[5],[5.5],[6],[6.5],[7],[7.5],
              [8], [8.5], [9], [9.5], [10]])

# EM algorithm using Gaussian Mixture Model
gmm = GaussianMixture(n_components=2, random_state=42)
gmm.fit(X)

# Cluster prediction
labels = gmm.predict(X)

print("Data:", X.flatten())
print("Clusters:", labels)
print("Means:", gmm.means_.flatten())
print("Weights:", gmm.weights_)

# Plot
plt.scatter(X, np.zeros_like(X), c=labels, s=100)
plt.xlabel("Data")
plt.title("Expectation-Maximization Algorithm")
plt.yticks([])
plt.show()
```
