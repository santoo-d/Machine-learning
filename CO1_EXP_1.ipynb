## FIND-S Algorithm Implementation

```python
import numpy as np

def find_s_algorithm(training_data):
    """
    Implements the FIND-S algorithm to find the most specific hypothesis.

    Args:
        training_data (list of lists): A list where each inner list represents a training example.
                                        The last element of each inner list is the target concept (label).
                                        Assumes that all positive examples are represented with 'Yes' or '1'.
                                        Attributes are categorical.

    Returns:
        list: The most specific hypothesis.
    """
    # Initialize the most specific hypothesis with the first positive example
    # The hypothesis will have '?' for unknown attributes and the actual attribute value
    # if it's consistent with all positive examples.

    hypothesis = []
    for example in training_data:
        if example[-1] == 'Yes' or example[-1] == 1:
            hypothesis = example[:-1]
            break

    if not hypothesis: # No positive examples found
        return "No positive examples to form a hypothesis."

    # Iterate through the remaining training examples
    for example in training_data:
        # Only consider positive examples
        if example[-1] == 'Yes' or example[-1] == 1:
            for i in range(len(hypothesis)):
                # If the hypothesis attribute does not match the example's attribute
                # generalize the hypothesis by replacing the attribute with '?'
                if hypothesis[i] != example[i]:
                    hypothesis[i] = '?'

    return hypothesis

# --- Demonstration ---

# Example Training Data
# Attributes: Sky, AirTemp, Humidity, Wind, Water, Forecast, EnjoySport
# Target Concept: EnjoySport (Yes/No)
training_data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

print("Training Data:")
for example in training_data:
    print(example)

# Find the most specific hypothesis
most_specific_hypothesis = find_s_algorithm(training_data)

print("\nMost Specific Hypothesis:")
print(most_specific_hypothesis)

# Another example with numeric labels and different attributes
training_data_2 = [
    ['Red', 'Small', 'Circle', 1],
    ['Blue', 'Medium', 'Square', 0],
    ['Red', 'Medium', 'Circle', 1],
    ['Green', 'Large', 'Triangle', 0],
    ['Red', 'Small', 'Square', 1]
]

print("\nTraining Data 2:")
for example in training_data_2:
    print(example)

most_specific_hypothesis_2 = find_s_algorithm(training_data_2)
print("\nMost Specific Hypothesis (Numeric Labels):")
print(most_specific_hypothesis_2)
```
