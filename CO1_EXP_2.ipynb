# Candidate-Elimination Algorithm Implementation

The Candidate-Elimination algorithm is used in machine learning to learn a concept from a set of training examples. It maintains a set of 'candidate' hypotheses that are consistent with the training data. This set is bounded by two sets: a set of maximally specific hypotheses (S) and a set of maximally general hypotheses (G).

Now, let's implement the `find_s` function to find the maximally specific hypothesis.

```python
def find_s(examples):
    # Initialize the most specific hypothesis h with the first positive example
    # Convert DataFrame row to list for easier processing
    specific_h = examples[0][0].tolist() if examples[0][1] == 'Yes' else None

    if specific_h is None:
        # Find the first positive example to initialize
        for i in range(len(examples)):
            if examples[i][1] == 'Yes':
                specific_h = examples[i][0].tolist()
                break
        if specific_h is None:
            return ['0'] * len(examples[0][0]) # No positive examples, all features are '0'

    num_attributes = len(specific_h)

    for i, (features, label) in enumerate(examples):
        features = features.tolist()
        if label == 'Yes':  # Only consider positive examples
            for j in range(num_attributes):
                if specific_h[j] != features[j]:
                    specific_h[j] = '?'
        # If label is 'No', ignore for find_s as it refines specific hypothesis
    return specific_h
```

Next, we implement the `find_g` function to find the maximally general hypothesis.

```python
def find_g(examples, specific_h):
    num_attributes = len(examples[0][0])
    # Initialize general hypotheses with all '?'s
    general_h = [['?' for _ in range(num_attributes)]]

    for i, (features, label) in enumerate(examples):
        features = features.tolist()
        if label == 'No':  # Only consider negative examples
            # For each negative example, prune general hypotheses
            # that are inconsistent with it.
            new_general_h = []
            for h in general_h:
                if not is_consistent(h, features): # If h covers the negative example, it's inconsistent
                    # Generate new, more specific hypotheses
                    for j in range(num_attributes):
                        if h[j] == '?':
                            if specific_h[j] != features[j]:
                                h_copy = list(h)
                                h_copy[j] = specific_h[j]
                                if not is_consistent(h_copy, features): # Ensure new hypothesis doesn't cover negative example
                                    new_general_h.append(h_copy)
                else:
                    new_general_h.append(h)
            general_h = new_general_h
        # Remove redundant hypotheses
        general_h = remove_redundant_hypotheses(general_h)
    return general_h

def is_consistent(hypothesis, example):
    for i in range(len(hypothesis)):
        if hypothesis[i] != '?' and hypothesis[i] != example[i]:
            return False
    return True

def remove_redundant_hypotheses(general_h):
    # Remove duplicate and more general hypotheses if a more specific one exists
    unique_hypotheses = []
    for h1 in general_h:
        is_redundant = False
        for h2 in general_h:
            if h1 != h2 and is_more_general(h2, h1):
                is_redundant = True
                break
        if not is_redundant and h1 not in unique_hypotheses:
            unique_hypotheses.append(h1)
    return unique_hypotheses

def is_more_general(h1, h2): # Checks if h1 is more general than or equal to h2
    for i in range(len(h1)):
        if h1[i] == '?' and h2[i] != '?':
            return False
        if h1[i] != '?' and h2[i] != '?' and h1[i] != h2[i]:
            return False
    return True
```

Finally, let's put it all together in the Candidate-Elimination algorithm.

```python
def candidate_elimination(X, y):
    # Combine features and labels into a list of (features, label) tuples
    examples = [(X.iloc[i], y.iloc[i]) for i in range(len(X))]

    specific_h = find_s(examples)
    print("Maximally Specific Hypothesis (S):", specific_h)

    general_h = find_g(examples, specific_h)
    print("Maximally General Hypotheses (G):", general_h)

    return specific_h, general_h

# Run the Candidate-Elimination algorithm
print("\nRunning Candidate-Elimination Algorithm...")
S, G = candidate_elimination(X, y)

print("\nFinal Version Space:")
print("S: ", S)
print("G: ", G)
```
